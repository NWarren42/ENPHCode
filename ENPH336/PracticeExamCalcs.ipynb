```python
import numpy as np
```

```python
# Constants
q = 1.602176565e-19  # C
k = 1.3806488e-23 # J/K
T = 300  # K
kT_q = k*T/q  # V (kT/q at 300K)
ni = 1e10  # cm^-3
V_applied = 0.4  # V

# P-side (minority carriers are electrons)
Na = 1e17  # cm^-3
mu_e_p = 800  # cm^2/Vs
tau_e_p = 0.1e-6  # s
D_e_p = mu_e_p * kT_q  # Diffusion constant for electrons on p-side
L_e_p = (D_e_p * tau_e_p)**0.5  # Diffusion length

print(f"P side electron diffusion constant: D = {D_e_p:.2e}\nP side electron diffusion length L = {L_e_p:.2e}")

# N-side (minority carriers are holes)
Nd = 1e15  # cm^-3
mu_h_n = 450  # cm^2/Vs
tau_h_n = 10e-6  # s
D_h_n = mu_h_n * kT_q  # Diffusion constant for holes on n-side
L_h_n = (D_h_n * tau_h_n)**0.5  # Diffusion length

print(f"N side hole diffusion constant: D = {D_h_n:.2e}\nN side hole diffusion length L = {L_h_n:2e}")

# Calculate saturation current density J0
J0 = q * ni**2 * (D_h_n / (L_h_n * Nd) + D_e_p / (L_e_p * Na))  # in A/cm^2

# Diode current density under forward bias
J = J0 * (np.exp(V_applied / kT_q) - 1)

print(f"Saturation current density J0 = {J0:.2e} A/cm^2")
print(f"Diode current density J = {J:.2e} A/cm^2")
```
