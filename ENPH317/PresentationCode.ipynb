```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
```

```python
def plotPMF(k_max, protons, prob_detect):
    k_vals = np.arange(0, k_max+1)
    y_vals = binom.pmf(k_vals, protons, prob_detect)
    
    plt.figure(figsize=(8, 4))
    plt.stem(k_vals, y_vals, basefmt=" ")
    plt.title(f"Binomial PMF (n={protons}, p={prob_detect})")
    plt.xlabel("Number of detected protons")
    plt.ylabel("Probability")
    plt.grid(True)
    plt.xlim(0, k_max)
    plt.ylim(0, np.max(y_vals)*1.1)
    plt.show()

    print(f"Most likely to see {k_vals[np.argmax(y_vals)]} protons at {np.max(y_vals)*100:.2f}% probability")

def plotCDF(k_max, protons, prob_detect):
    k_vals = np.arange(0, k_max+1)
    y_vals = binom.cdf(k_vals, protons, prob_detect)
    
    plt.figure(figsize=(8, 4))
    plt.step(k_vals, y_vals, where='post')
    plt.title(f"Binomial CDF (n={protons}, p={prob_detect})")
    plt.xlabel("Number of detected protons")
    plt.ylabel("Cumulative Probability")
    plt.grid(True)
    plt.xlim(0, k_max)
    plt.ylim(0, 1.001)
    plt.show()
```

```python
# Example usage:
plotPMF(10, 60, 0.01)
plotCDF(10, 60, 0.01)
```

```python
plotPMF(40, 24*60, 0.01)
plotCDF(40, 24*60, 0.01)
```

```python
plotPMF(10000, 360*24*60, 0.01)
plotCDF(10000, 360*24*60, 0.01)
```
